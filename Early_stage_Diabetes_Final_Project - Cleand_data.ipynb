### Introduction to Early Stage Diabetes Risk Prediction
Diabetes is one of the most common chronic diseases worldwide, affecting millions of people. It occurs when the body cannot produce enough insulin or cannot use insulin effectively, resulting in high blood glucose levels. Early detection of diabetes is important because it helps individuals take preventive measures, receive timely medical care, and reduce the risk of serious health complications.

The purpose of this analysis is to understand the key factors associated with early-stage diabetes risk and present meaningful insights through clear and effective data visualizations. This project uses Python libraries such as **Pandas, NumPy, Plotly, Matplotlib, and Seaborn** to analyze a diabetes dataset containing patient information, signs, and symptoms related to diabetes. The analysis aims to identify important patterns and relationships in the data, explore common diabetes-related symptoms, and understand the factors that may indicate an increased risk of developing diabetes. The insights obtained through data analysis and visualization can support early risk assessment and promote awareness of the importance of timely medical screening.

### (3)Creating Derived Columns

### Firts column:Age Group
This will help you analyze how diabetes classification varies across different age groups.

```python

# Create Age Groups
age = [0, 29, 44, 59, float('inf')]
group = ['Young', 'Adult', 'Middle-aged', 'Senior']

df['Age Group'] = pd.cut(df['Age'],bins=age,labels=group,include_lowest=True)

# Display the results
print(df[['Age', 'Age Group']].head(10))

# Count patients in each age group
print(df['Age Group'].value_counts().sort_index())
```

### Second Column : Symptom Count

```python
#create column for symptom count
# List of symptom columns
symptom_columns = [
    'Polyuria',
    'Polydipsia',
    'sudden weight loss',
    'weakness',
    'Polyphagia',
    'Genital thrush',
    'visual blurring',
    'Itching',
    'Irritability',
    'delayed healing',
    'partial paresis',
    'muscle stiffness',
    'Alopecia',
    'Obesity'
]

# Count the Yes responses in each row
df['Symptom Count'] = (df[symptom_columns].apply(lambda col: col.astype(str).str.strip().str.lower().eq('yes')).sum(axis=1))

print(df[['Age', 'Symptom Count', 'class']].head(10))
```
