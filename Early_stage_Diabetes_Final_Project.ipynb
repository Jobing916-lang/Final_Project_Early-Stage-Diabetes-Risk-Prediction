### Introduction to Early Stage Diabetes Risk Prediction
Diabetes is one of the most common chronic diseases worldwide, affecting millions of people. It occurs when the body cannot produce enough insulin or cannot use insulin effectively, resulting in high blood glucose levels. Early detection of diabetes is important because it helps individuals take preventive measures, receive timely medical care, and reduce the risk of serious health complications.

The purpose of this analysis is to understand the key factors associated with early-stage diabetes risk and present meaningful insights through clear and effective data visualizations. This project uses Python libraries such as **Pandas, NumPy, Plotly, Matplotlib, and Seaborn** to analyze a diabetes dataset containing patient information, signs, and symptoms related to diabetes. The analysis aims to identify important patterns and relationships in the data, explore common diabetes-related symptoms, and understand the factors that may indicate an increased risk of developing diabetes. The insights obtained through data analysis and visualization can support early risk assessment and promote awareness of the importance of timely medical screening.

### 8.Frequency of Reported Symptoms

***Visualization Type:*** Horizontal Bar Chart (Matplotlib / Seaborn)

***Purpose:*** To analyze, rank, and compare the total frequency of individual symptoms reported by patients across the entire dataset.

***Insight:*** The bar chart clearly identifies weakness as the most common symptom in the dataset, reported by nearly 160 individuals. It is closely followed by Polyuria, Itching, and delayed healing, all exceeding 120 occurrences. Conversely, Obesity is the least frequently reported clinical feature, appearing in fewer than 50 individuals. ```python

```python
symptom_columns = [
    "Polyuria",
    "Polydipsia",
    "sudden weight loss",
    "weakness",
    "Polyphagia",
    "Genital thrush",
    "visual blurring",
    "Itching",
    "Irritability",
    "delayed healing",
    "partial paresis",
    "muscle stiffness",
    "Alopecia",
    "Obesity"
]

# Count Yes responses, ignoring letter case and extra spaces
yes_counts = df[symptom_columns].apply(
    lambda col: col.astype("string")
    .str.strip()
    .str.casefold()
    .eq("yes")
    .sum()
).sort_values()

plt.figure(figsize=(10, 7))
yes_counts.plot(kind="barh")

plt.title("Frequency of Reported Diabetes Symptoms")
plt.xlabel("Number of Individuals Reporting Yes")
plt.ylabel("Symptoms")
plt.tight_layout()
plt.show()
```

### 9.Correlation Heatmap

***Visualization Type:*** Heatmap (Seaborn heatmap)

***Purpose:*** To evaluate, quantify, and visualize the linear correlation coefficients between pairs of clinical variables, demographic attributes, and diabetes classes.

***Insight:*** The heatmap highlights strong linear dependencies using a diverging color scale. Notably, class_Positive shares its strongest positive correlations with Polyuria_Yes (0.62) and Polydipsia_Yes (0.59), indicating that excessive urination and extreme thirst are powerful predictive indicators of a positive diagnosis. Conversely, a perfect negative correlation (-1.00) correctly shows up between mutually exclusive one-hot encoded pairs (such as Gender_Female vs. Gender_Male), serving as a data validation checkpoint.

```python

# Select relevant variables
heatmap_columns = [
    "Age",
    "Gender",
    "Polyuria",
    "Polydipsia",
    "sudden weight loss",
    "weakness",
    "Polyphagia",
    "Symptom Count",
    "class"
]

heatmap_df = df[heatmap_columns].copy()

# Encode categorical columns
heatmap_df = pd.get_dummies(
    heatmap_df,
    columns=[
        "Gender",
        "Polyuria",
        "Polydipsia",
        "sudden weight loss",
        "weakness",
        "Polyphagia",
        "class"
    ],
    dtype=int
)

corr = heatmap_df.corr(numeric_only=True)

plt.figure(figsize=(12, 9))
sns.heatmap(
    corr,
    cmap="coolwarm",
    center=0,
    annot=True,
    fmt=".2f"
)

plt.title("Correlation Heatmap of Selected Variables")
plt.tight_layout()
plt.show()
```

#### 10. Diabetes Class Proportion by Polyuria

***Visualization Type:*** Box plot · Seaborn

***Purpose:*** To compare, evaluate, and visualize the distribution, central tendency, and overall spread of concurrent symptom counts across positive and negative diabetes diagnostic classes.

***Insight:*** The box plot highlights a clear upward shift in symptom density for individuals diagnosed with diabetes. Notably, the Positive class exhibits a higher median symptom count of 7, with the middle 50% of patients (Interquartile Range) falling between 5 and 9 symptoms. In contrast, the Negative class maintains a lower median symptom count of 4, with its typical range spanning from 1 to 6 symptoms. While both groups include individuals presenting zero symptoms, the significantly higher overall distribution and a maximum whisker reaching 13 symptoms in the Positive group indicate that an accumulation of multiple overlapping symptoms is a strong clinical indicator of a positive diagnosis.

```python

# Calculate class proportions for Polyuria Yes/No
polyuria_data = pd.crosstab(
    df["Polyuria"],
    df["class"],
    normalize="index"
) * 100

# Order categories when both are present
preferred_order = ["No", "Yes"]
order = [
    value for value in preferred_order
    if value in polyuria_data.index
]
order += [
    value for value in polyuria_data.index
    if value not in order
]

polyuria_data = polyuria_data.reindex(order)

fig = px.line(
    polyuria_data.reset_index().melt(
        id_vars="Polyuria",
        var_name="Diabetes Class",
        value_name="Percentage"
    ),
    x="Polyuria",
    y="Percentage",
    color="Diabetes Class",
    markers=True,
    title="Diabetes Class Proportion by Polyuria",
    labels={
        "Polyuria": "Polyuria Response",
        "Percentage": "Individuals (%)"
    }
)

fig.show()
```
